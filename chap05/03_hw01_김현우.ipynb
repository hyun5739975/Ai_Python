```python
# 문제2) 다음 딕셔너리가 있다.

# scores = {
#     "Kim": 80,
#     "Lee": 92,
#     "Park": 68,
#     "Choi": 100
# }

# 다음을 출력하시오.

# - 평균 점수
# - 최고 점수 학생
# - 80점 이상 학생 이름

scores = {
    "Kim": 80,
    "Lee": 92,
    "Park": 68,
    "Choi": 100
}

list_text1 = []
list_text2 = []
list_text3 = []
total = 0

for key in scores : 
  list_text1.append(scores[key])
  list_text2.append(key)
  
  total += scores[key]

  if scores[key] >= 80:
    list_text3.append(key)

avg1 = total / len(scores)

max1 = max(list_text1)
max2 = list_text1.index(max1)
max3 = list_text2[max2]

  
print("80점 이상의 학생 이름 : ", list_text3)
print("평균 : {}".format(avg1))
print("최고점수 학생 {}의 점수 : {}점".format(max3, max1))
```
