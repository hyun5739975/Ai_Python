```python
# 가변 매개변수 함수
# 함수 매개변수 정의하는데 * 추가시 가변
def print_n_times(n, *value) :
  # n번 반복
  for i in range(n) :
    for values in value:
      print(values)
      pass

# 함수를 호출합니다.
print_n_times(3, "안녕하세요", "즐거운", "파이썬 프로그래밍")
print_n_times(2, "서울시", "마포구", "인창동")


# 가변 매개변수 뒤에는 일반 매개변수가 올 수 없다.
# 가변 매개변수는 하나만 사용할 수 있다. -> 파이썬 프로그래밍 특성상 어디까지인지 모름
def print_n_times_2(n, *values1) :
  # n번 반복
  for i in range(n) :
    for values in values1:
      print(values)

print_n_times_2(2, "아시아", "usa", "japan")

```

```python
# 기본 매개변수 (ex) n=2 대입연산자가 없으면 일반 매개변수
def print_n_times(value="테스트1", n=2): # 기본 n 값은 2로 고정
  print(n)

  for i in range(n):
    print(value, i)


# 함수를 호출합니다.
#print_n_times("안녕하세요", 5)
#print_n_times("안녕하세요") # 호출 시 기본 매개변수가 있으면 해당 부분을 생략 가능
#print_n_times(n =3, value="안녕하세요")
print_n_times(value="테스트2")

# 기본 매개변수(default argument) 뒤에는 일반 매개변수(Non-default argument)가 올 수 없다.
# 변수에는 다양한 데이터 타입을 담을 수 있으므로 혼돈을 방지하기 위해 막아버림
# 단 초기화를 해주면 가능
# def print_n_times(n=2, value) :
def print_n_times(n=2, value="") :
  for i in range(n):
    print(value)

print_n_times(value="Test")
```

```python
# 키워드 매개변수 기본 매개변수는 가변 매개변수 앞에 오면 안된다 모호함 때문에
# 가변 매개변수 대입할 값 갯수 상관없이 담을 수 있음
def print_n_times(*values, n=2):
  for i in range(n):
    for value in values:
      print(i, value)
  print()

# 함수를 호출합니다.
# print_n_times("asdasda","asdadasd","asdasdasd", 2) # 에러
print_n_times("asdasda","asdadasd","asdasdasd", n=4)

```

```python
# 기본 매개변수 중에서 필요한 값만 입력하기
# 여러 함수 호출 형태

def test(a, b=10, c=100) :
  print(f"[{a}] + [{b}]+ [{c}] = {a+b+c}")

# 1) 기본 형태
test(50, 20, 30)

# 2) 키워드 매개변수로 모든 매개변수를 지정한 형태
test(50, 20, 30)

# 3) 키워드 매개변수로 모든 매개변수를 마구잡이로 지정한 형태
test (c=300, a=100, b=20)

# 4) 키워드 매개변수로 일부 매개변수만 지정한 형태
test(50, b=30)
```

```python
# 범위 내부의 정수를 모두 더하는 함수

# 함수를 선언합니다.
def sum_all(start, end) :
  output = 0

  for i in range(start, end+1): # def range(start=0, end, step=1)
    output += i

  return output

# 함수를 호출합니다.
print("0 to 100:", sum_all(0, 100))
print("0 to 1000:", sum_all(0, 1000))
print("50 to 100:", sum_all(50, 100))
print("500 to 1000:", sum_all(500, 1000))
```

```python
# 기본 매개변수와 키워드 매개변수를 활용해 범위의 정수를 더하는 함수

# 함수를 선언합니다.
def sum_all(start=0, end=100, step=1) :
  output = 0

  for i in range(start, end+1, step):
    output += i

  return output

# 함수를 호출합니다.
print("A.", sum_all(20, 50 ,1))
print("B.", sum_all(10, 1000, 2))
print("C.", sum_all(100, 500, 1))
print("D.", sum_all(end=50))
print("E.", sum_all(end=50, step = 2))
```
