```python
# 함수의 매개변수로 함수 전달하기

# 매개변수로 받은 함수를 10번 호출하는 함수 -> 매개변수에 함수를 전달 할 수 있다.
# func = print_hello -> func() 호출 = print_hello 함수 호출
def call_10_times(func):
    for i in range(10):
        func()


# 간단한 출력하는 함수
def print_hello():
    print("안녕하세요\n")

# 조합하기
call_10_times(print_hello)
```

```python
# map() 함수와 filter() 함수

# 함수를 선언합니다.
def power(item):
    return item * item
def under_3(item):
    return item < 3

# 변수를 선언합니다.
list_input_a = [1, 2, 3, 4, 5]

# map() 함수를 사용합니다. (함수, 데이터 전달) -> 특정 이벤트가 발생되면 우리가 처리하게끔 기능을 구현 call back 함수
output_a = map(power, list_input_a)
print("# map() 함수의 실행 결과")
print("map(power, list_input_a):", output_a)
print("map(power, list_input_a):", list(output_a))
print()

# filter() 함수를 사용합니다. -> 필터 해당 데이터 제외하고 나머지 데이터 걸러줌
output_b = filter(under_3, list_input_a) # 매개변수 함수, 데이터
print("# filter() 함수의 실행 결과")
print("filter(under_3, list_input_a):", output_b)
print("filter(under_3, list_input_a):", list(output_b))
```

```python
# 람다
# def power(item):
#     return item * item

# 람다 함수를 선언합니다. (위 처럼 정의 안해도 간단하게 할 수 있음) -> 간단한 로직일 경우 주로 사용
power = lambda item: item * item
under_3 = lambda item: item < 3

# 변수를 선언합니다.
list_input_a = [1, 2, 3, 4, 5]

# map() 함수를 사용합니다.
output_a = map(power, list_input_a)
print("# map() 함수의 실행 결과")
print("map(power, list_input_a):", output_a)
print("map(power, list_input_a):", list(output_a))
print()

# filter() 함수를 사용합니다.
output_b = filter(under_3, list_input_a)
print("# filter() 함수의 실행 결과")
print("filter(under_3, list_input_a):", output_b)
print("filter(under_3, list_input_a):", list(output_b))
```

```python
# 인라인 람다 따로 변수를 만들지 않고 map 이나 filter 에 직접 정의
# 인라인 = 다이렉트

# 변수를 선언합니다.
list_input_a = [1, 2, 3, 4, 5]

# map() 함수를 사용합니다.
output_a = map(lambda x: x * x, list_input_a)
print("# map() 함수의 실행 결과")
print("map(power, list_input_a):", output_a)
print("map(power, list_input_a):", list(output_a))
print()

# filter() 함수를 사용합니다.
output_b = filter(lambda x: x < 3, list_input_a)
print("# filter() 함수의 실행 결과")
print("filter(under_3, list_input_a):", output_b)
print("filter(under_3, list_input_a):", list(output_b))
```

```python
# 랜덤하게 1000명의 키와 몸무게 만들기

# 랜덤한 숫자를 만들기 위해 가져옵니다.
import random
# 간단한 한글 리스트를 만듭니다.
hanguls = list("가나다라마바사아자차카타파하")
print(hanguls)
    
# 파일을 쓰기 모드로 엽니다.
with open("info.txt", "w", encoding="utf-8") as file:
    for i in range(1000):
        # 랜덤한 값으로 변수를 생성합니다.
        name = random.choice(hanguls) + random.choice(hanguls)
        weight = random.randrange(40, 100)
        height = random.randrange(140, 200)
        # 텍스트를 씁니다.
        file.write("{}, {}, {}\n".format(name, weight, height))
```

```python
# 반복문으로 파일 한 줄씩 읽기

with open("info.txt", "r") as file:
    for line in file:
        # 변수를 선언합니다.
        (name, weight, height) = line.strip().split(", ")

        # 데이터가 문제없는지 확인합니다: 문제가 있으면 지나감
        if (not name) or (not weight) or (not height):
            continue

        # 결과를 계산합니다.
        bmi = int(weight) / ((int(height) / 100) ** 2)
        result = ""
        if 25 <= bmi:
            result = "과체중"
        elif 18.5 <= bmi:
            result = "정상 체중"
        else:
            result = "저체중"

        # 출력합니다.
        print('\n'.join([
            "이름: {}",
            "몸무게: {}",
            "키: {}",
            "BMI: {}",
            "결과: {}"
        ]).format(name, weight, height, bmi, result))
        print()
```

```python
# 제너레이터
# 이터레이터를 직접 만들때 사용하는 코드 (yield 사용) -> next() 함수 여러개의 데이터를 하나씩 꺼내올수 있음 이터레이터, 반복 객체를 대상으로 객체 (이터러블)
# 일반 함수와 달리 호출해도 함수 내부코드가 실행되지 않음

# 함수를 선언합니다.
def test():
    print("함수가 호출되었습니다.")
    yield test

# 함수를 호출합니다.
print("A 지점 통과")
test()

print("B 지점 통과")
test()
print(test())
```

```python
# 제너레이터 객체와 next() 함수호출

# 함수를 선언합니다.
def test():
    print("A 지점 통과")
    yield 1 # 이터레이터
    print("B 지점 통과")
    yield 2
    print("C 지점 통과")
    yield 3
    # yield 4

	
# 함수를 호출합니다.
output = test()
	
# next() 함수를 호출합니다.
print("D 지점 통과")
a = next(output)
print(a)
	
# print("E 지점 통과")
b = next(output)
print(b)

print("F 지점 통과")
c = next(output)
print(c)

# 한번 더 실행하기 yield 키워드 및 내부 데이터 없으면 에러남
#next(output)
```

```python
# 키워드 매개변수에 함수 전달하기

books = [{
    "제목": "만화로 보는 5분 한국사",
    "가격": 16920
}, {
    "제목": "니체의 초월자",
    "가격": 15210
}, {
    "제목": "품격 있는 대화를 위한 지식 브리핑",
    "가격": 22500
}]

def select_price(book):
    return book["가격"]

print("# 가장 저렴한 책")
print(min(books, key=select_price)) # map, filter 함수랑 비슷 = > min, max 콜백함수들을 정의해서 함께 제공을 제공해줘야함
print(min) # 정수형 데이터는 해도 됨, 딕셔너리 데이터인 경우 모호성 때문에 위 코드문으로 해줘야함 

print("# 가장 비싼 책")
print(max(books, key=select_price))
```

```python
# 콜백 함수를 람다로 바꾸기

books = [{
    "제목": "만화로 보는 5분 한국사",
    "가격": 16920,
    "연도": 2022
}, {
    "제목": "니체의 초월자",
    "가격": 15210,
    "연도": 2021
}, {
    "제목": "품격 있는 대화를 위한 지식 브리핑",
    "가격": 22500,
    "연도": 2020
}]

print("# 가장 저렴한 책")
print(min(books, key=lambda book: book["가격"]))
print()

print("# 가장 비싼 책")
print(max(books, key=lambda book: book["가격"]))
```

```python
# 딕셔너리 오름차순 정렬하기

books = [{
    "제목": "만화로 보는 5분 한국사",
    "가격": 16920
}, {
    "제목": "니체의 초월자",
    "가격": 15210
}, {
    "제목": "품격 있는 대화를 위한 지식 브리핑",
    "가격": 22500
}]

print("# 가격 오름차순 정렬")
books.sort(key=lambda book: book["가격"]) # sort = 오름차순
for book in books:
    print(book)
```
