```python
## 문제1. 구구단 출력 프로그램을 작성해 보세요.

#2 ~ 9단 타이틀
print("2단 \t\t  3단 \t\t  4단 \t\t  5단 \t\t  6단 \t\t  7단 \t\t  8단 \t\t  9단")

# 1부터 9까지 행을 반복
for num in range(9):

    # 2단 ~ 9 단 까지의 열을 반복 
    for dan in range(2, 10):
        print("{} * {} = {}".format(dan, num+1, dan * (num+1)), end="\t") # 단 * 행 => 곱셈값 * \t 이스케이프로 여백 맞춤
    
    #줄바꿈    
    print("\n")  # 다음단으로 갈땨 이스케이프 문자로 줄바꿈 

```

```python
##문제2. 로또 번호 생성기를 프로그램으로 작성해 보세요.

# time 함수 가져오기
import time

# 빈 리스트 추가
lotto = []

# 리스트(lotto) 데이터갯수가 10개일 때까지 반복
while len(lotto) < 10:
    lotto_num = int(time.time() * 100000 % 100000) # 타임함수로 데이터를 담아 변수에 담기
    
    num = lotto_num % 45 # 최대 20번까지 숫자를 담음
    
    if lotto.count(num) == 0: # 로또 리스트에 0개 인 경우만 
    
        lotto.append(num) # 리스트에 추가    

    else : # 리스트 안에 데이터가 존재 시
        pass # 루프 종료
print("오늘의 로또번호: {}".format(lotto)) # 출력

```

```python
## 문제3. numbers = [12, 7, 25, 9, 18, 30] 
## 합계 / 평균 / 최댓값 / 최솟값 / 짝수 개수 / 홀수 개수를 sum(), max() 등을 사용하지 않고 직접 반복문으로 구현해보세요.

numbers     = [12, 7, 25, 9, 18, 30] 
total       = 0
num         = 0
data_max    = numbers[0]
data_min    = numbers[0]
count_1     = 0
count_2     = 0


# 데이터의 갯수 반복문
for num_list in range(len(numbers)) : 
    num_list += 1
print("데이터의 갯수 : ({})".format(num_list))
print("")

# 각 계산식 반복문
for i in numbers : 
    num += 1
    
    # 합계
    total += i

    # 평균
    avg = total / num_list
    
    print("{}의 번째 값 : ({})".format(num, i))
    
    # 최대값 & 최소값
    if i > data_max : 
        data_max = i
    elif i < data_min :
        data_min = i

    # 짝수 & 홀수 갯수
    if i % 2 == 0 :
        count_1 += 1
    elif i % 2 == 1:
        count_2 += 1 
        

# 결과값 출력
# 1) 합계
print()
print("리스트 데이터 합계 : ({})".format(total))

# 2) 평균
print()
print("데이터의 평균 값 : ({})".format(avg))

# 3) 최대값
print("최대값 : ({})".format(data_max))

# 4) 최소값
print("최소값 : ({})".format(data_min))

# 5) 짝수 개수
print("짝수갯수 : ({})".format(count_1))

# 6) 홀수 개수
print("홀수 갯수 : ({})".format(count_2))

```
