```python
# 런타임 오류(Runtime error)/예외(Exception) - 프로그램이 실행되는 과정에서 발생하는 오류

# 프로그램 시작
print("# 프로그램이 시작되었습니다.")

# 예외 발생 코드
# list_a = [1, 2, 3, 4, 5]

print(list_a[1])
```

```python
# 조건문으로 예외 처리하기

# 숫자를 입력받습니다.
user_input_a = input("정수 입력> ")

# 사용자 입력이 숫자로만 구성되어 있을 때
if user_input_a.isdigit(): # 숫자형일때만 True를 반환
    # 숫자로 변환합니다.
    number_input_a = int(user_input_a)
    # 출력합니다.
    print("원의 반지름:", number_input_a)
    print("원의 둘레:"  , 2 * 3.14 * number_input_a)
    print("원의 넓이:"  , 3.14 * number_input_a * number_input_a)
else:
    print("정수를 입력하지 않았습니다.")
```

```python
# try except 구문으로 예외를 처리합니다.

try : # 기능 부분
  # 숫자를 입력받습니다.
  user_input_a = input("정수 입력> ")

  # 숫자로 변환합니다.
  number_input_a = int(user_input_a)
  # 출력합니다.
  print("원의 반지름:", number_input_a)
  print("원의 둘레:", 2 * 3.14 * number_input_a)
  print("원의 넓이:", 3.14 * number_input_a * number_input_a)    
except: # 에러 처리 부분
  print("정수를 입력하지 않았습니다.")
```

```python
# 숫자로 변환되는 것들만 리스트에 넣기

# 변수를 선언합니다.
list_input_a = ["52", "273", "32", "스파이", "103"]

# 반복을 적용합니다.
list_number = []
for item in list_input_a: #이터레이터 객체를 통해서 해당 데이터를 읽어 반환을 함
    # 숫자로 변환해서 리스트에 추가합니다.
    try:
        float(item) # 실수형
        list_number.append(item) # 데이터 추가
    except: # 한줄이상의 코드문이 필요함 ex) pass 아무것도 하지마라
        pass

# 출력합니다.
print("{} 내부에 있는 숫자는".format(list_input_a))
print("{}입니다.".format(list_number))
```

```python
# try ~ except ~ else 구문으로 예외를 처리합니다.

try : # 정상적인 코드 (이벤트 발생부분)
    # 숫자를 입력받습니다.
    user_input_a = input("정수 입력> ")

    # 숫자로 변환합니다.
    number_input_a = int(user_input_a)

except : # try 구문에서 발생된 이벤트 선정해 해당 코드문을 수행함
    print("정수를 입력하지 않았습니다.")

else : # 예외가 발생되지 않은 경우 코드문이 수행됨
    print("원의 반지름:", number_input_a)
    print("원의 둘레:", 2 * 3.14 * number_input_a)
    print("원의 넓이:", 3.14 * number_input_a * number_input_a)    
```

```python
# finally 구문
try : # 정상적인 코드 (이벤트 발생부분)
    # 숫자를 입력받습니다.
    user_input_a = input("정수 입력> ")

    # 숫자로 변환합니다.
    number_input_a = int(user_input_a)

except : # try 구문에서 발생된 이벤트 선정해 해당 코드문을 수행함
    print("정수를 입력하지 않았습니다.")

else : # 예외가 발생되지 않은 경우 코드문이 수행됨
    print("예외가 발생되지 않음")  
    print("원의 반지름:", number_input_a)
    print("원의 둘레:", 2 * 3.14 * number_input_a)
    print("원의 넓이:", 3.14 * number_input_a * number_input_a)  

finally : # 이벤트 상관없이 무조건 수행함
  print("일단 프로그래밈 끝났습니다.")
  print("정상적으로 처리가 되듼 예외가 상관없이 처리 끝났습니다.")
```

```python
# test() 함수를 선언합니다.
def test():
    print("test() 함수의 첫 줄입니다.")
    try:
        print("try 구문이 실행되었습니다.")
        return # 호출한곳에 복귀하라
        #print("try 구문의 return 키워드 뒤입니다.")
    except:
        print("except 구문이 실행되었습니다.")
    else:
        print("else 구문이 실행되었습니다.")
    finally: # 함수 호출 시 return으로 복귀할때도 무조건 수행됨
        print("finally 구문이 실행되었습니다.")
    print("test() 함수의 마지막 줄입니다.")

# test() 함수를 호출합니다.
test()
```

```python
# 반복문과 함께 사용하는 경우
# finally 반복문 내 break로 반복문 마지막 한번은 수행

print("프로그램이 시작되었습니다.")

while True:
    try:
        print("try 구문이 실행되었습니다.")
        break
        print("try 구문의 break 키워드 뒤입니다.")
    except:
        print("except 구문이 실행되었습니다.")
    finally:
        print("finally 구문이 실행되었습니다.")
    
    print("while 반복문의 마지막 줄입니다.")

print("프로그램이 종료되었습니다.")
```
