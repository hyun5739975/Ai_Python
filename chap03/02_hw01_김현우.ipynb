# 과제 3

문제1. 
피타고라스의 정리
     a^2 + b^2 = c^2 에서
사용자로 부터 밑변의 길이와, 높이의 값을 입력 받았을 때, 
빗변의 길이를 출력하는 프로그램을 작성해 보세요.

문제2. 
정수 3개를 입력받아(input() 함수 이용) 가장 큰 수를 출력하세요.


문제3. 
연도를 입력받아 윤년인지 판별하세요.

윤년 조건:
- 400으로 나누어 떨어지면 윤년
- 또는 4로 나누어 떨어지고 100으로는 나누어 떨어지지 않으면 윤년

문제4. 
사용자에게 태어난 연도의 값을 입력받아, 띠를 출력하는 프로그램을 작성하세요.


```python
# 문제1 빗변의 길이를 출력하는 프로그램

# 밑변의 길이
# 밑변의 길이를 입력 받음
bottom_lenth = input("밑변의 길이 : ")
print(bottom_lenth)

# 밑변의 높이
# 밑변의 높이를 입력 받음
bottom_height = input("밑변의 높이 : ")
print(bottom_height)

# 피타고라스 정리 각 값의 ** 제곱 연산자 사용해서 계산, int형, float형으로 형변환, [0.5 제곱 (구글링검색)]
hypotens_lenth = ((int(bottom_lenth) ** 2) + (int(bottom_height) ** 2)) ** 0.5

# .format()함수로 결과값 도출
print("계산된 결과값 : {} 입니다.".format(hypotens_lenth))
```

```python
# 문제2 값 3개를 입력 받아 가장 큰 수를 출력하세요

# num1 를 입력 받음
num1 = input("숫자 1 : ")
print(num1)

# num2 를 입력 받음
num2 = input("숫자 2 : ")
print(num2)

# num3 를 입력 받음
num3 = input("숫자 3 : ")
print(num3)

if(num1 >= num2 or num1 >= num3) :
  # 가장 큰 수 출력
  print("가장 큰 수 num1 : {} 입니다.".format(num1))
elif(num2 >= num1 or num2 >= num3) :
  # 가장 큰 수 출력
  print("가장 큰 수 num2 : {} 입니다.".format(num2))
elif(num3 >= num1 or num3 >= num2 ) :
    # 가장 큰 수 출력
  print("가장 큰 수 num3 : {} 입니다.".format(num3))
```

```python
# 문제3. 
# 연도를 입력받아 윤년인지 판별하세요.

# 윤년 조건:
# - 400으로 나누어 떨어지면 윤년 [조건 1]
# - 또는[or] 4로 나누어 떨어지고 100으로는 나누어 떨어지지 않으면 윤년 [조건 2]

# num_year 를 입력 받음
num_year = input("연도 : ")
print(num_year)
print(int(num_year) % 400)
print(int(num_year) % 4)
print(int(num_year) % 100)

if((int(num_year) % 400 == 0) or (int(num_year) % 4 == 0 and int(num_year) % 100 != 0)) : 
  print("해당 연도 {}년 => 윤년입니다.".format(num_year))
else :
  print("해당 연도 윤년이 아닙니다.")


```

```python
# 문제4. 
# 사용자에게 태어난 연도의 값을 입력받아, 띠를 출력하는 프로그램을 작성하세요.

# my_year 를 입력 받음
my_year = input("출생연도 : ")
print("출생연도 > {} 입니다.".format(my_year))

# 2026년생		말띠
# 2025년생		뱀띠
# 2024년생		용띠
# 2023년생		토끼띠
# 2022년생		호랑이띠
# 2021년생		소띠
# 2020년생		쥐띠
# 2019년생		돼지띠
# 2018년생		개띠
# 2017년생		닭띠
# 2016년생		원숭이띠
# 2015년생		양띠

## 2026 ~ 2015 년도
i_flags = [ "2026"
           ,"2025"
           ,"2024"
           ,"2023"
           ,"2022"
           ,"2021"
           ,"2020"
           ,"2019"
           ,"2018"
           ,"2017"
           ,"2016"
           ,"2015"]

## 12간지 종류
j_flags = [ "말띠"
           ,"뱀띠"
           ,"용띠"
           ,"토끼띠"
           ,"호랑이띠"
           ,"소띠"
           ,"쥐띠"
           ,"돼지띠"
           ,"개띠"
           ,"닭띠"
           ,"원숭이띠"
           ,"양띠"]

## 해당 년도(key), 12종 띠(value) -> zip함수 사용하여 각 매개변수 key, value를 대입
for key, value in zip (i_flags, j_flags) :
  if((key) == my_year) : # 조건문 연도 값과 해당 년도 값이 일치 할 경우
    print("당신 띠는 > {} 입니다.".format(value)) ## 해당 값을 출력 value(띠)
    break
```
