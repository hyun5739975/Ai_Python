### * 날짜/시간 활용하기

```python
# 날짜/시간과 관련된 기능을 가지고옴
import datetime

# 현재 날짜/시간을 구합니다.
dateTime = datetime.datetime.now()

# 출력합니다.
print("{}년 {}월 {}일 {}시간 {}분 {}초".format(dateTime.year
                                             , dateTime.month
                                             , dateTime.day
                                             , dateTime.hour
                                             , dateTime.minute
                                             , dateTime.second))

result = [dateTime.year        
        , dateTime.month
        , dateTime.day
        , dateTime.hour
        , dateTime.minute
        , dateTime.second ]

print("result ", result)
```

```python
# 날짜/시간과 관련된 기능을 가지고옴
import datetime



# 현재 날짜/시간을 구합니다.
now = datetime.datetime.now()
print(now, "월 입니다.")

month_01 = int(input("해당 월을 입력하세요"))

# 봄 구분 (3~5)
if(3 <= now.month <= 5 ) :
  print("해당 {}월은 봄입니다".format(now.month))

# 여름 구분 (6~8)
if(6 <= now.month <= 8 ) :
  print("해당 {}월은 여름입니다".format(now.month))

# 가을 구분 (9~11)
if(9 <= now.month <= 11 ) :
  print("해당 {}월은 가을입니다".format(now.month))

# 겨울 구분 (12~2)
if(1 <= now.month <= 2 or now.month == 12 ) :
  print("해당 {}월은 겨울입니다".format(now.month))
```
