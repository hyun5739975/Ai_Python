# 클래스의 추가적인 구문

```python
# isinstance() 함수

# 학생 클래스를 선언합니다.
class Student:
    def study(self):
        print("공부를 합니다.")

# 선생님 클래스를 선언합니다.
class Teacher:
    def teach(self):
        print("학생을 가르칩니다.")

# 교실 내부의 객체 리스트를 생성합니다.
classroom = [Teacher(), Student(), Student(), Student(), Student()]

# 반복을 적용해서 적절한 함수를 호출하게 합니다.
# 서로 다른 데이터형 있으면 하나씩 꺼내올 수 있게 하는 것을 isintance로 수행가능 
for person in classroom: 
    if isinstance(person, Teacher): # 맞으면 true 첫번째가 teacher가 맞는지
        person.teach()
    elif isinstance(person, Student):
        person.study()
```

```python
#  __str__() 함수
# 이터러블 : 2개 이상 데이터 하나씩 꺼내올수있음 리스트, 튜플, 딕셔너리, 문자열 > 이터러블 
# 이터레이터 : 실제로 값을 하나씩 꺼냄
# 클래스를 선언합니다.
# 오브젝트, 오버라이딩 상속
class Student: 
    def __init__(self, name, korean, math, english, science):
        self.name    = name
        self.korean  = korean
        self.math    = math
        self.english = english
        self.science = science

    def get_sum(self):
        return self.korean + self.math + self.english + self.science

    def get_average(self):
        return self.get_sum() / 4

    def __str__(self): # 오버라이딩
         return f"{self.name}\t {self.get_sum()}\t {self.get_average()}"


# 학생 리스트를 선언합니다.
students = [
    Student("유재석", 87, 98, 88, 95), # 공간 할당
    Student("신동엽", 92, 98, 96, 98),
    Student("강호동", 76, 96, 94, 90),
    Student("박명수", 98, 92, 96, 92),
    Student("김구라", 95, 98, 98, 98),
    Student("김국진", 64, 88, 92, 92),
]

# 출력합니다.
print("이름", "총점", "평균", sep="\t")
for student in students:
    print(str(student))
```

```python
# 클래스 함수

# 클래스를 선언합니다.
class Student: # 자료형은 리스트
    # 클래스 변수
    count = 0 # 클래스에서 메모리 할당 (자동)
    students = [] # 리스트 자료형 클래스에서 메모리 할당 (자동)

    # 클래스 함수 (데코레이터로 처리가 되어있다 classmethod) (자동 모듈 수행이 될때마다 메모리 할당)
    @classmethod
    def print(cls):
        print("---학생 목록----")
        print("이름\t총점\t평균")
        for student in cls.students:
            print(str(student)) # 해당 데이터에 대한 데이터를 출력
        print("----------------")


    # 인스턴스 함수 호출할때마다 메모리 할당 -> self에 시작주소 append 저장 후 전달-> 생성자 함수 호출
    def __init__(self, name, korean, math, english, science):
        self.name    = name
        self.korean  = korean
        self.math    = math
        self.english = english
        self.science = science
        Student.count += 1 # 메모리 할당과 상관없음
        Student.students.append(self) # 기능을 호출

    def get_sum(self):
        return self.korean + self.math +\
            self.english + self.science

    def get_average(self):
        return self.get_sum() / 4

    def __str__(self):
        return "{}\t{}\t{}".format(\
            self.name,\
            self.get_sum(),\
            self.get_average())

# 학생 리스트를 선언합니다.
Student("유재석", 87, 98, 88, 95) # 1번 수행후
Student("신동엽", 92, 98, 96, 98) # class 데이터 수 파악 후 2번 수행
Student("강호동", 76, 96, 94, 90)
Student("박명수", 98, 92, 96, 92)
Student("김구라", 95, 98, 98, 98)
Student("김국진", 64, 88, 92, 92)
Student("서장훈", 82, 86, 98, 88)
Student("이경규", 88, 74, 78, 92)
Student("정형돈", 97, 92, 88, 95)
Student("이수근", 45, 52, 72, 78)

# 현재 생성된 학생을 모두 출력합니다.
Student.print()
```

```python
# private 변수와 getter/setter

# 모듈을 가져옵니다.
import math

# 클래스를 선언합니다. -> 잘못된 값을 처리하려면 클래스 외부에서 클래스를 사용할때 직접적 변경 못하게끔 정의
class Circle:
  def __init__(self, radius): # 생성자 함수 
    self.radius = radius # 메모리 공간 데이터 = 매개변수 데이터
  def get_circumference(self) : # 둘레
    return 2 * math.pi * self.radius # 2 * 원주율 * 반지름
  def get_area(selef): # 넓이
    return math.pi * (selef.radius ** 2) # (self.radius * self.radius)
  
# 원의 둘레와 넓이를 구합니다. 
circle = Circle(10) # 생성자 함수 호출 (메모리 할당) 
circle.radius = -2                               # Circle 클래스의 radius 속성에 음수를 넣는다면? = 둘레는 음수값이 나오면 안됨
print("원의 둘레:", circle.get_circumference())  # 음수값 발생
print("원의 넓이:", circle.get_area())
```

```python
# 데코레이터를 사용한 게터와 세터

# 모듈을 가져옵니다.
import math

# 클래스를 선언합니다.
class Circle:
    def __init__(self, radius):
        self.__radius = radius
    def get_circumference(self):
        return 2 * math.pi *  self.__radius
    def get_area(self):
        return math.pi * (self.__radius ** 2)

    # 게터와 세터를 선언합니다.
    # 파이썬에선 게터를 만들지 않으면 세터를 만들 수 없다.
    # 데코레이터 문법
    @property # getter로 동작
    def radius(self):
         return self.__radius

    @radius.setter # setter로 동작
    def radius(self, value):
        if value <= 0:
            raise TypeError("길이는 양의 숫자여야 합니다.") # 만약 음수값이 전달 된 경우
        self.__radius = value


# 원의 둘레와 넓이를 구합니다.
print("# 데코레이터를 사용한 Getter와 Setter")
circle = Circle(10)
print("원래 원의 반지름: ", circle.radius)
circle.radius = -2
print("변경된 원의 반지름: ", circle.radius)
print()

# 강제로 예외를 발생시킵니다.
print("# 강제로 예외를 발생시킵니다.")
# circle.radius = -10
```

```python
# 상속
# 클래스를 정의 데이터를 관리하는 경우 많이 없음 
# 부모 클래스를 선언합니다. Parent 모든 요소 -> child 안에 추가 관리 할 수 있음
class Parent:
    def __init__(self):
        self.value = "테스트"
        print("Parent 클래스의 __init()__ 메서드가 호출되었습니다.")
    def test(self):
        print("Parent 클래스의 test() 메서드입니다.")

# 자식 클래스를 선언합니다. Parent -> Child 상속
class Child(Parent):
    pass

# 자식 클래스의 인스턴스를 생성하고 부모의 메서드를 호출합니다.
child = Child()
child.test()
print(child.value)
```

```python
# 상속의 활용 - 예외 클래스 만들기

class CustomException(Exception):
    def __init__(self):
        super().__init__() #클래스를 상속하는 구조안에서 할 수 있음 자식 클래스는 부모클래스를 쳐다봐라 (exception) 호출해주는 함수 (의무적으로 1번은 호출)
        # 부모에 변수가 있을 경우 전달할게 없는 경우 부모에 전달할 데이터가 없으면 에러가 나오니 자식생성자 함수에서 super(부모 생성자함수 전달) 변수값을 초기화
raise CustomException
```

```python
# 자식 클래스로써 부모의 함수 재정의(오버라이딩)하기

class CustomException(Exception):
    def __init__(self):
        super().__init__()
        print("##### 내가 만든 오류가 생성되었어요! #####")

    # overriding 함수 정의 -> 재정의
    def __str__(self):
        return "오류가 발생하였습니다."


raise CustomException
```

```python
# 자식 클래스에 새로운 함수 정의하기

# 사용자 정의 예외를 생성합니다.
class CustomException(Exception):
    def __init__(self, message, value):
        super().__init__()
        self.message = message
        self.value = value

    def __str__(self):
        return self.message

    def print(self):
        print("###### 오류 정보 ######")
        print("메시지:", self.message)
        print("값:", self.value)

# 예외를 발생시켜 봅니다.
try : 
    age = int(input("사용자 입력 나이 : "))

    if age < 0 :
        raise CustomException("나이의 값이 이상함", age)

    elif age > 0 :
        print("정상 동작")

except CustomException as e:
    e.print()
```
