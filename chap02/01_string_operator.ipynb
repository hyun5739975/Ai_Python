# 자료형 (Data Type)
### - Escape Sequence Character

```python
# " \" " => "표시
print("철수가 \"안녕\" 이라고 인사를 했어")

# " \' " => '표시
print("철수가 \'안녕\' 이라고 인사를 했어")

# " \n " => line 줄바꿈
print("철수가 \n안녕\n이라고 인사를 했어")

# " \t " => tab 사이 공백
print("철수가 \t안녕\t이라고 인사를 했어")

# " \r " => 이스케이프 앞 문자열 지우기
print("철수가 \r안녕\r이라고 인사를 했어")

# " \\ " => 역슬레쉬 표시
print("C:\\workspaces\\python")
```

```python
# \ : 역슬레시 여백없애기
print(
      """\
      안녕하세요 
      철수가 놀러 왔어요.
      기쁜 마음에 반갑습니다.
      인사를 했어요 
      """\
    )
```

```python
# \" : 긴 문자열 안에 "" 표시
print(
      """
      안녕하세요 
      \"철수\"가 놀러 왔어요.
      기쁜 마음에 반갑습니다.
      인사를 했어요 
      """
    )
```

```python
### - 문자열 연산자
print("안녕"+"하세요") #여백 없음
print("안녕","하세요") #여백 있음
print("안녕하세요 제 나이는 "+str(30)+"세 입니다.") #문자와 숫자 형변환
```

```python
### - 문자열 반복 연산자 * 
print(3*5)
print("안녕하세요" * 3)
```

```python
# 문자 선택 연산자(인덱싱) : [] => 문자열 한 문자 -> 문자열 범위 초과시 에러 // string index out of range
test = "안녕하세요"
a = 0
for a in range(5): print(test[a])
print()
for a in range(5): print(test[-a])
```

```python
# 문자열 범위 선택 연산자(슬라이싱) : [시작위치, 증감값] -> 시작위치 생략가능
print("안녕하세요"[0:2])
print("안녕하세요"[1:3])
print("안녕하세요"[3:4])
print("안녕하세요"[:4])
print("안녕하세요"[1:])
```

```python
# lenth 문자열 길이
print(type(len))
print("문자열길이 : ",len("안녕하세요. "))
```
