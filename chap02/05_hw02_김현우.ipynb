```python
# p29 실습 : 사직 연산
data1 = 5
data2 = 7
print("=====================")

print("p29 실습")
print(str(data1)+" + "+str(data2)+" = ", data1 + data2) 
print(str(data1)+" + "+str(data2)+" = ", data1 - data2)
print(str(data1)+" + "+str(data2)+" = ", data1 * data2)
print(str(data1)+" + "+str(data2)+" = ", data1 / data2)

print("=====================")

# p30 실습 : 정수 나눗셈 
print("p30 실습")
print("3 / 2 = ", 3 / 2) # 나눗셈
print("3 // 2 = ", 3 // 2) # 나눗셈을 정수형태로 변환
print("5 % 2 = ", 5 % 2) # 두값을 나눌때 나머지 값

print("=====================")

# p31 실습 제곱연산
print("p31 실습")
print("2 ** 1 =", 2 ** 1) # 2의 1승
print("2 ** 2 =", 2 ** 2) # 2의 2승
print("2 ** 3 =", 2 ** 3) # 2의 3승
print("2 ** 4 =", 2 ** 4) # 2의 4승

print("=====================")

# p32 실습
print("p32 실습")
print(2 + 2 - 2 * 2 / 2 * 2) # 4 - 2 * 2
print(2 - 2 + 2 / 2 * 2 + 2) # 0 + 1 * 4

print("=====================")

# p33 실습
print("p33 실습")
result= (5+3) * 2
print(result) # -> 8 * 2

result= 5+(3 * 2)
print(result) # -> 5 + 6

print("=====================")

# p34 실습 : 타입에러
print("p34 실습")
string = "문자열"
number = 273
#result = string + number # Eror code : can only concatenate str (not "int") to str
#print(result)

result2 = string+" : "+str(number)
#해결방법 (숫자->문자열로 데이터 형변환)

print(result2)

print("=====================")

# p35 실습 : 연산자 우선순위
print("p35 실습")
# 괄호 > 곱셈 = 나눗셈 > 덧셈 = 뺄셈
print("안녕"+"하세요" * 3)
# 괄호 > 곱셈 = 나눗셈 > 덧셈 = 뺄셈
print(("안녕"+"하세요") * 3)
# 괄호 > 곱셈 = 나눗셈 > 덧셈 = 뺄셈
print("안녕"+("하세요" * 3))
```
