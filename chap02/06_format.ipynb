## 4. 숫자와 문자열의 다양한 기능
### - 문자열의 format() 함수

```python
# format() 함수로 숫자를 문자열로 변환
string_a = "{}".format(10) #숫자형 -> 문자열화
# 출력
print(type(string_a), string_a) 

string_b = "{}".format("10") #문자열 -> 숫자화
print(type(string_b), string_b) 
```

```python
# won = input("가격을 입력하세요 > ")
# total = "{}".format(won)
# print("가격의 값은 : " + total + " 원입니다.")

# won = input("가격을 입력하세요 > ")
# total = "가격의 값은 : {} 원입니다.".format(won)
# print(total)

# format_a = "{}, {}, {}, {}, {}".format(1,3.14,"asdasd", True, False)
# print(format_a)

# 정수
output_a = "{:d}".format(52)

# 특정 칸에 출력하기
output_b = "{:5d}".format(52)   # 5칸
output_c = "{:10d}".format(52)  # 10칸

# 빈칸에 0으로 채우기
output_d = "{:05d}".format(52)  # 양수
output_e = "{:05d}".format(-52) # 음수

print("표현식 응용")
print(output_a)
print(output_b)
print(output_c)
print(output_d)
print(output_e)

# 기호화 함께 출력하기
output_f = "{:+5d}".format(52)  # 양수
output_g = "{:+5d}".format(-52) # 음수
output_h = "{: d}".format(52)   # 양수 : 기호 부분 공백
output_i = "{: d}".format(-52)  # 음수 : 기호 부문 공백

print("# 기호화 함께 출력하기")
print(output_f)
print(output_g)
print(output_h)
print(output_i)

# 조합하기
output_j = "{:+5d}".format(52)    # 기호를 뒤로 밀기 : 양수
output_k = "{:+5d}".format(-52)   # 기호를 뒤로 밀기 : 음수
output_l = "{:=+d}".format(52)    # 기호를 뒤로 얖으로 밀기 : 양수
output_m = "{:=+d}".format(-52)   # 기호를 뒤로 앞으로기 : 음수
output_n = "{:+05d}".format(52)   # 0으로 채우기 : 양수
output_o = "{:+05d}".format(-52)  # 0으로 채우기 : 음수

print("# 조합하기")
print(output_j)
print(output_k)
print(output_l)
print(output_m)
print(output_n)
print(output_o)

# float 실수 자료형 기본
output_a1 = "{:+f}".format(52.273)  
output_b1 = "{:15f}".format(52.273)    # 15칸 만들기
output_c1 = "{:+15f}".format(52.273)   # 15칸 부호 추가하기
output_d1 = "{:+015f}".format(52.273)  # 15칸에 부호 추가하고 0으로 채우기

print("# float 실수 자료형 기본")
print(output_a1)
print(output_b1)
print(output_c1)
print(output_d1)

# 소수점 아래 자릿수 지정
output_a2 = "{:15.3f}".format(52.273)  
output_b2 = "{:15.2f}".format(52.273)    
output_c2 = "{:15.1f}".format(52.273)   

print("# 소수점 아래 자릿수 지정")
print(output_a2)
print(output_b2)

# 소수점 제거
output_a3 = 52.0  
output_b3 = "{:g}".format(output_a3)    

print("# 소수점 제거")
print(output_a3)
print(output_b3)

# 값을 출력하고 싶으면 print함수를 자주 이용
# 파이썬은 "{:d}".format(52) 이런 형태로 디버깅하면 print 함수 없이 출력은 가능함 단 마지막 형태의 코드만 출력가능
```
