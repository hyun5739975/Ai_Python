## 자료형 (DataType)
### · 정수형
### · 실수형

```python
print(375)
print(3.14)
print("정수형 : ", type(375))
print("실수형 : ", type(3.14))
```

```python
# 덧셈 + -> 숫자형 더하기
print("5 + 7   = ", 5 + 7)
# 뺄셈 - -> 숫자형 빼기
print("8 - 2   = ", 8 - 2)
# 곱셈 * -> 숫자형 곱하기
print("2 * 8   = ", 2 * 8)
# 나눗셈 / -> 숫자형 나누기
print("10 / 2  = ", 10 / 2)
# 나눗셈 결과값 실수형 -> 몫
print("10 / 2 = ", 10 // 2)
# 나머지 % -> 숫자형 나머지
print("17 % 3  = ", 17 % 3)
```

```python
# 괄호 > 곱셈 = 나눗셈 > 덧셈 = 뺄셈
print("안녕"+"하세요" * 3)
# 괄호 > 곱셈 = 나눗셈 > 덧셈 = 뺄셈
print(("안녕"+"하세요") * 3)
# 괄호 > 곱셈 = 나눗셈 > 덧셈 = 뺄셈
print("안녕"+("하세요" * 3))
```
