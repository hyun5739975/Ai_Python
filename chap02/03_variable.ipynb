```python
# 변수 선언과 할당
pi = 3.141592                                     #원주율
# r = 12.5                                        #반지름

r = input("반지름의 값을 입력하세요 > ")

print("원주율 => ", pi)                           # 원주율
print("반지름 => ", r)                            # 반지름

# 원의 둘레, 원의 넓이 계산
d_result1 = 2 * pi * float(r)                     # 원의 둘레
d_result2 = pi * float(r) * float(r)              # 원의 넓이

print("원의 둘레 => "+ str(d_result1)+" 입니다.") # 원의 둘레
print("원의 넓이 => "+ str(d_result2)+" 입니다.") # 원의 넓이

print("==========================")
iflags = ["원의 둘레", "원의 넓이"]
jflags = [d_result1, d_result2]

for key, value in zip(iflags,jflags) :            #for = 반복문 / key(매개변수1), value(매개변수2) = zip(변수1, 변수2) 
  print (f"{key} => {value}" + " 입니다")         #f = format(포맷) / => 변수와 매개변수 갯수, 순서가 똑같아야함 key(ifalgs) : value(jflags)
```

## - input() 함수
#### 사용자가 어느때에 직접 값을 입력할 수 있도록 도와주는 함수

```python
# 입력을 받습니다.
name = input("이름을 입력하세요 : ")
print("안녕하세요 "+ name + "입니다.")

test1 = input("테스트1 : ")
print(type(test1))
print("테스트 1 값은 :", test1)

test2 = input("테스트2 : ")
print(type(test2))
print("테스트 2 값은 :", test2)

test3 = input("연산자를 입력하세요")
print("연산자 값 : ", test3)

test4 = input("데이터형태 입력하세요")
if(test4 =="int") :
  print("데이터 값 : ", test4)

elif(test4 =="float") :
  print("데이터 값 : ", test4)

else :
  print("지원하지 않는 데이터입니다.")

if(test3 == "+" and test4 =="int") : 
  print("덧셈 결과값 : ", int(test1) + int(test2), "입니다.")
elif(test3 == "+" and test4 =="float") : 
  print("덧셈 결과값 : ", float(test1) + float(test2), "입니다.")

if(test3 == "-" and test4 =="int") : 
  print("뺄셈 결과값 : ", int(test1) - int(test2), "입니다.")
elif(test3 == "-" and test4 =="float") : 
  print("뺄셈 결과값 : ", float(test1) - float(test2), "입니다.")

if(test3 == "*" and test4 =="int") : 
  print("곱셈 결과값 : ", int(test1) * int(test2), "입니다.")
elif(test3 == "*" and test4 =="float") : 
  print("곱셈 결과값 : ", float(test1) * float(test2), "입니다.")

if(test3 == "/" and test4 =="int") : 
  print("나눗셈 결과값 : ", int(test1) / int(test2), "입니다.")
elif(test3 == "/" and test4 =="float") : 
  print("나눗셈 결과값 : ", float(test1) * float(test2), "입니다.")

elif (test4 != "int" and test4 != "float"):
  print("데이터 타입이 올바르지 않습니다.")


```
